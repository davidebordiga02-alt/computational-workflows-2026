1. What is nf-core?

nf-core is a community project that provides ready-to-use, standardized bioinformatics pipelines built with Nextflow. They are reproducible, follow FAIR principles, and run anywhere using containers like Docker.

2. How many pipelines are there currently in nf-core?

more than 150 (157)

3. Are there any non-bioinformatic pipelines in nf-core?

Yes. for example for proteomics, metabolomics, imaging, and general data processing. 

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A pipeline is a series of connected computational steps that automatically process data.
(Each step takes an input, performs a task and passes its output to the next step.)

We use pipelines to analyze large datasets in a reproducible, scalable, and automated way.

5. Why do you think nf-core adheres to strict guidelines?

To ensure that all pipelines are high-quality, reproducible, and interoperable. Without them, each pipeline could produce inconsistent results. Guidelines also make it easier for the community to contribute, review, and maintain pipelines. 

6. What are the main features of nf-core pipelines?

The main features are reproducibility, portability, scalability, modularity, standardization, containerization (Docker/Singularity)

## Lets look at the results

### What is differential abundance analysis?

Differential abundance analysis is a statistical approach to identify features (genes, proteins, metabolites, etc.) whose abundance differs significantly between two or more conditions or groups.

Give the most important plots from the report:

```python
import os
import glob
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# 👉 INSERISCI QUI IL PERCORSO DELLA CARTELLA DEL REPORT
base_dir = r"/Users/tesbaro/Desktop/CompWorkflow/computational-workflows-2026/day_01/results/differentialabundance3/report"

# Parole chiave per identificare i plot importanti
keywords = {
    "Enrichment plot: HALLMARK_KRAS_SIGNALING_DN": "enplot_HALLMARK_KRAS_SIGNALING_DN",
    "Ranked Gene List Correlation Profile": "ranked_list_corr",
    "Random ES distribution: HALLMARK_KRAS_SIGNALING_DN": "gset_rnd_es_dist",
    "Heatmap top 50 features": "heat_map_1",
    "Butterfly plot": "butterfly_plot"
}

# Cerca ricorsivamente tutti i PNG nella cartella
all_pngs = glob.glob(os.path.join(base_dir, "**", "*.png"), recursive=True)

print(f"Trovati {len(all_pngs)} file PNG in {base_dir}\n")
for f in all_pngs:
    print(f)

# Seleziona i file corrispondenti alle parole chiave
selected = []
for title, kw in keywords.items():
    matches = [f for f in all_pngs if kw in os.path.basename(f)]
    if matches:
        selected.append((title, matches[0]))   # prende il primo match
    else:
        print(f"⚠️ Nessun file trovato per: {title} (keyword: {kw})")

# Visualizza in griglia
if selected:
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    axes = axes.flatten()
    for ax, (title, filename) in zip(axes, selected):
        img = mpimg.imread(filename)
        ax.imshow(img)
        ax.set_title(title, fontsize=10)
        ax.axis('off')
    for i in range(len(selected), len(axes)):
        axes[i].axis('off')
    plt.tight_layout()
    plt.show()
else:
    print("Nessun plot trovato. Controlla i nomi dei file o il percorso.")
```

Global level: There is evidence of differential abundance between hND6 and mCherry (heatmap separation, ranked list bias, butterfly plot separation).

Gene-set level: The specific set HALLMARK_KRAS_SIGNALING_DN is not significantly enriched (p = 0.513, FDR = 0.614). You cannot conclude that this pathway is differentially abundant.
