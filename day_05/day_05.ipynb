### 1. Describe the concept of Workflows, Processes and Channels we deal with in Nextflow

Workflows: is the pieline structure, a collection of processes connected by channels. It receives input, passes data through the process graph, and produces output files. It's declarative: you define connections, Nextflow handles scheduling. allow process to work in parallel.



Processes: computational unit that takes input from channels, runs a script (bash/Python/R), and emits output to channels. Each run executes in its own work/ directory. Can use containers for reproducibility. Key directives: cpus, memory, errorStrategy, container.






Channels: the connections between processes, asynchronous queues that carry data from process outputs to process inputs.

### Why was CONVERTTOUPPER run so often?

The 2 rows in samplesheet_2.csv each produced 2 chunk files (due to block_size splitting), so 4 total CONVERTTOUPPER executions.
1. Each input string was split into 2 chunks (block_size 4 for "Hello World" -> "Hell", "o Wo", "rld"; block_size 3 for "Computational Workflows" -> "Com", "put", "ati", "ona", "l W", "ork", "flo", "ws")
2. Each chunk becomes a separate file processed by CONVERTTOUPPER
3. The pipeline processes all chunks in parallel, so 2 rows × 2 chunks = 4 process executions
4. This is expected behavior - Nextflow parallelizes by default
