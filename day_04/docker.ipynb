### Explain the RUN and ENV lines you added to the file

RUN: Updates the package list, installs curl and cowsay, then cleans the apt cache to keep the image small. Runs at build time (creates a layer in the image).<br>
ENV: Adds /usr/games to the PATH so cowsay (installed in /usr/games/) can be run directly without the full path. Runs at runtime (available in the container environment).
